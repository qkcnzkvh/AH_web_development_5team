## 1. 각 디렉터리의 역할

디렉터리란?  
쉽게 말해서 폴더를 의미한다.

### 1.1 `app/core/`

프로젝트 전체에서 공통으로 사용하는 설정 및 핵심 기능을 관리하는 디렉터리

- `config.py`: 환경 변수 및 프로젝트 설정 관리
- `db/databases.py`: 데이터베이스 연결 설정 (엔진·세션·Base)
- `db/models.py`: 모델 공통 컬럼(Mixin) 정의
- `security.py`: 인증 및 보안 관련 기능

### 1.2 `app/models/`

데이터베이스에 저장될 데이터의 구조를 정의하는 디렉터리

- SQLAlchemy 모델 클래스 작성
- 테이블 이름 및 컬럼 정의
- 컬럼의 데이터 타입 및 관계 설정

### 1.3 `app/repositories/`

데이터베이스에 접근하여 데이터를 조회하거나 변경하는 코드를 작성하는 디렉터리

- 데이터 생성 (INSERT)
- 데이터 조회 (SELECT)
- 데이터 수정 (UPDATE)
- 데이터 삭제 (DELETE)

### 1.4 `app/schemas/`

API 요청 및 응답 데이터의 형식을 정의하는 디렉터리

- 요청 데이터 검증
- 필수 필드 및 데이터 타입 지정
- API 응답 데이터 형식 정의
- 주로 Pydantic 사용

### 1.5 `app/services/`

서비스의 비즈니스 로직을 처리하는 디렉터리

- 비즈니스 규칙 처리
- Repository를 이용한 데이터 조회 및 저장
- 여러 데이터베이스 작업 조합
- 입력값에 따른 처리 로직 작성

### 1.6 `app/apis/`

클라이언트의 HTTP 요청을 처리하는 API 엔드포인트를 작성하는 디렉터리

- API 경로 정의
- HTTP 메서드 지정 (`GET`, `POST`, `PUT`, `DELETE`)
- 요청 데이터 전달
- Service 호출 및 응답 반환

## 1·2번 보충 ①. 전체 처리 흐름

일반적인 요청 처리 흐름은 다음과 같다.

1. **APIs**: 클라이언트의 요청을 받는다.
2. **Schemas**: 요청 데이터의 형식을 검증한다.

`core`는 공통 설정 및 핵심 기능을 제공하며, `main.py`는 애플리케이션을 구성하고 각 API 라우터를 연결한다.

## 3. 데이터베이스 연결 설정 (`app/core/db/databases.py`)

**역할:** 애플리케이션과 MySQL 사이의 연결을 만들고 관리하는 파일입니다. 접속 정보는 `app/core/config.py`의 `settings`에서 가져오고, 이 파일에서 **엔진 · 세션 팩토리 · Base · 세션 제공 함수**를 만들어 repositories/services 등 다른 계층이 사용합니다.

### 3-0. 전체 연결 흐름

| 단계 | 파일 | 하는 일 |
|---|---|---|
| ① | `.env` | DB 접속 정보(`DB_USER`, `DB_PASSWORD`, `DB_HOST`, `DB_PORT`, `DB_NAME`)를 저장. **직접 생성 필요**, Git에는 올리지 않음 |
| ② | `docker-compose.yml` | 같은 `.env` 값으로 MySQL 컨테이너를 생성 (`${DB_NAME}` 등) |
| ③ | `app/core/config.py` | 같은 `.env`를 읽어 `settings` 객체로 제공 |
| ④ | `app/core/db/databases.py` | `settings`로 접속 URL을 만들고 엔진/세션/Base 생성 |
| ⑤ | `app/apis/...` | `Depends(async_get_db)`로 세션을 받아 DB 사용 |

즉 **하나의 `.env`를 docker-compose(MySQL 생성)와 config.py(앱 접속)가 함께 사용**하기 때문에 두 쪽의 접속 정보가 항상 일치합니다.

### 3-1. 설정 파일 (`app/core/config.py`)

```python
from pydantic_settings import BaseSettings


class Settings(BaseSettings):
    DB_USER: str = "root"
    DB_PASSWORD: str = "password1234"
    DB_HOST: str = "localhost"
    DB_PORT: str = "3306"
    DB_NAME: str = "ai_health"

    model_config = {
        "env_file": ".env",
        "extra": "ignore"
    }


settings = Settings()
```

| 코드 | 설명 |
|---|---|
| `BaseSettings` | pydantic-settings의 클래스. 환경변수/`.env` 값을 읽어 타입 검증까지 해줌 |
| `DB_USER` ~ `DB_NAME` | DB 접속 정보. 코드의 기본값(`= "root"` 등)은 `.env`나 환경변수에 값이 **없을 때만** 사용됨 |
| `"env_file": ".env"` | 프로젝트 루트의 `.env` 파일에서 값을 읽음 (`.env` 값이 기본값보다 우선) |
| `"extra": "ignore"` | `.env`에 Settings에 없는 변수(예: `DB_ROOT_PASSWORD`)가 있어도 에러 없이 무시 |
| `settings = Settings()` | 앱 전체에서 `from app.core.config import settings`로 가져다 쓰는 단일 객체 |

**`.env` 예시** (docker-compose와 공유하므로 `DB_ROOT_PASSWORD`도 필요)

```env
DB_ROOT_PASSWORD=루트비밀번호
DB_USER=root
DB_PASSWORD=실제비밀번호
DB_HOST=localhost
DB_PORT=3306
DB_NAME=ai_health
```

> 현재 `DB_PASSWORD` 기본값이 코드에 적혀 있어요. 실제 비밀번호는 `.env`에만 두고 `.gitignore`에 `.env`를 추가하는 것이 안전합니다.

### 3-2. DB 연결 (`app/core/db/databases.py`)

```python
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.ext.asyncio import AsyncSession, create_async_engine, async_sessionmaker
from typing import AsyncGenerator
from app.core.config import settings


DATABASE_PREFIX = "mysql+asyncmy://"
DATABASE_URI = f"{settings.DB_USER}:{settings.DB_PASSWORD}@{settings.DB_HOST}:{settings.DB_PORT}/{settings.DB_NAME}"
DATABASE_URL = f"{DATABASE_PREFIX}{DATABASE_URI}"

# 비동기 엔진 생성
async_engine = create_async_engine(DATABASE_URL, echo=False, future=True)
# 비동기 세션 팩토리 생성
AsyncSessionLocal = async_sessionmaker(bind=async_engine, autoflush=False, expire_on_commit=False)
# 모델 베이스 생성
Base = declarative_base()

# 세션 생성 함수
async def async_get_db() -> AsyncGenerator[AsyncSession, None]:
    async with AsyncSessionLocal() as db:
        yield db
```

### 3-3. 코드 설명

**① DB URL 조립**

| 코드 | 설명 |
|---|---|
| `DATABASE_PREFIX = "mysql+asyncmy://"` | `DB종류+드라이버://` 형식. MySQL을 **asyncmy**(비동기 드라이버)로 연결한다는 뜻 |
| `DATABASE_URI` | `유저:비밀번호@호스트:포트/DB이름` 부분을 `settings` 값으로 조립 |
| `DATABASE_URL` | 최종 접속 주소. 예: `mysql+asyncmy://root:pw@localhost:3306/ai_health` |

**② 엔진 / 세션 / Base**

| 요소 | 설명 |
|---|---|
| `create_async_engine(...)` | DB와의 실제 연결(커넥션 풀)을 관리하는 **비동기 엔진**. `echo=False`는 SQL 로그를 출력하지 않음 (디버깅 시 `True`) |
| `async_sessionmaker(...)` | 요청마다 `AsyncSession`을 만들어내는 **세션 팩토리** |
| `autoflush=False` | 쿼리 실행 전에 변경사항을 자동으로 DB에 보내지 않음 (필요할 때 직접 `flush`/`commit`) |
| `expire_on_commit=False` | commit 후에도 객체 속성을 만료시키지 않음. 비동기에서는 만료된 속성에 접근하면 자동 재조회가 안 되어 에러가 나므로 **사실상 필수 설정** |
| `declarative_base()` | 모든 모델이 상속받는 부모 클래스. **Alembic이 이 `Base.metadata`를 보고 테이블 변경을 감지** |

**③ `async_get_db()` (의존성 주입용 세션 제공 함수)**

- FastAPI에서 `Depends(async_get_db)`로 사용하면, 요청이 들어올 때 세션을 열어 라우터/서비스에 전달하고 요청이 끝나면 `async with`가 **자동으로 세션을 닫아줍니다**.
- 자동 commit은 아니므로, 저장·수정 시 repository/service에서 `await db.commit()`을 직접 호출해야 합니다.

```python
# 사용 예시 (app/apis/...)
from fastapi import Depends
from sqlalchemy.ext.asyncio import AsyncSession
from app.core.db.databases import async_get_db

@router.get("/users")
async def get_users(db: AsyncSession = Depends(async_get_db)):
    ...
```

**④ 개선하면 좋은 점 (선택)**

| 현재 코드 | 권장 | 이유 |
|---|---|---|
| `from sqlalchemy.ext.declarative import declarative_base` | `from sqlalchemy.orm import declarative_base` | SQLAlchemy 2.0부터 앞의 경로는 deprecated |
| `future=True` | 제거 | 2.0에서는 기본 동작이라 불필요 |
| 비밀번호를 URL에 그대로 삽입 | `urllib.parse.quote_plus(settings.DB_PASSWORD)` | 비밀번호에 `@`, `/` 같은 특수문자가 있으면 URL이 깨짐 |

## 4. SQLAlchemy 모델 작성 및 Alembic 마이그레이션 방법

### 4-1. SQLAlchemy 모델 작성

**역할:** DB 테이블 구조를 파이썬 클래스로 정의합니다. 클래스 하나가 테이블 하나, 속성 하나가 컬럼 하나에 대응됩니다. 모델은 3번에서 만든 `Base`를 상속받아 작성합니다.

**이 프로젝트의 모델 관련 파일 위치**

| 위치 | 내용 |
|---|---|
| `app/core/db/models.py` | 여러 테이블에서 **공통으로 쓰는 컬럼 묶음(Mixin)** 정의 (실제 테이블은 아님) |
| `app/models/` | **실제 테이블 모델**(예: `User`)을 작성하는 폴더 |
| `app/models/__init__.py` | 모델을 import 해두는 곳. `alembic/env.py`가 `from app import models`로 불러오므로 **여기서 import 되어야 Alembic이 테이블을 인식** |

### 4-1-1. 공통 Mixin (`app/core/db/models.py`)

```python
import uuid as uuid_pkg
from datetime import UTC, datetime

from sqlalchemy import Boolean, DateTime, text
from sqlalchemy.dialects.mysql import CHAR
from sqlalchemy.orm import Mapped, mapped_column
from uuid6 import uuid7


class UUIDMixin:
    uuid: Mapped[uuid_pkg.UUID] = mapped_column(
        CHAR(36), primary_key=True, default=uuid7
    )


class TimestampMixin:
    created_at: Mapped[datetime] = mapped_column(
        DateTime, default=datetime.now(UTC), server_default=text("current_timestamp(0)")
    )
    updated_at: Mapped[datetime | None] = mapped_column(
        DateTime, nullable=True, onupdate=datetime.now(UTC), server_default=text("current_timestamp(0)")
    )


class SoftDeleteMixin:
    deleted_at: Mapped[datetime | None] = mapped_column(DateTime, nullable=True)
    is_deleted: Mapped[bool] = mapped_column(Boolean, default=False)
```

| Mixin | 추가되는 컬럼 | 설명 |
|---|---|---|
| `UUIDMixin` | `uuid` (CHAR(36), PK) | 기본키를 숫자 id 대신 **UUID v7**(시간순 정렬이 가능한 UUID)로 사용. `default=uuid7`로 INSERT 시 자동 생성 |
| `TimestampMixin` | `created_at`, `updated_at` | 생성/수정 시각. `server_default=current_timestamp(0)`은 DB가 직접 채워주는 기본값 |
| `SoftDeleteMixin` | `deleted_at`, `is_deleted` | 행을 실제로 지우지 않고 **삭제 표시만** 하는 소프트 삭제용 컬럼 |

> Mixin은 그 자체로는 테이블이 되지 않고, 모델 클래스가 **상속받아 컬럼을 물려받는** 용도입니다.

**개선하면 좋은 점 (선택)**

| 현재 코드 | 권장 | 이유 |
|---|---|---|
| `default=datetime.now(UTC)` | `default=lambda: datetime.now(UTC)` | 괄호를 붙여 호출하면 **서버가 시작된 시점의 시간이 고정값으로 저장**됨. `onupdate`도 동일 |
| `default=uuid7` (UUID 객체) + `CHAR(36)` | `default=lambda: str(uuid7())` | CHAR 컬럼에는 문자열로 넣는 것이 안전 (드라이버에 따라 UUID 객체 바인딩 에러 가능) |

### 4-1-2. 실제 모델 작성 예시 (`app/models/user.py`)

아래는 Mixin을 사용하는 **예시**입니다. 과제에서 사용하는 실제 테이블에 맞게 컬럼을 바꿔서 쓰세요.

```python
# app/models/user.py
from sqlalchemy import String
from sqlalchemy.orm import Mapped, mapped_column

from app.core.db.databases import Base
from app.core.db.models import UUIDMixin, TimestampMixin, SoftDeleteMixin


class User(UUIDMixin, TimestampMixin, SoftDeleteMixin, Base):
    __tablename__ = "users"

    email: Mapped[str] = mapped_column(String(255), unique=True, nullable=False)
    name: Mapped[str] = mapped_column(String(100), nullable=False)
```

- 상속 순서는 **Mixin들 → `Base`** 순서로 씁니다.
- 위 모델이 만드는 컬럼: `uuid`, `created_at`, `updated_at`, `deleted_at`, `is_deleted`(Mixin) + `email`, `name`(직접 정의)

**모델 등록 (`app/models/__init__.py`)**

```python
from app.models.user import User
```

> 이 import가 없으면 `alembic revision --autogenerate`가 모델을 찾지 못해 **빈 마이그레이션**이 만들어집니다.

### 4-2. Alembic이란?

모델 변경 사항(테이블 추가, 컬럼 수정 등)을 **버전 관리하며 DB에 반영**해주는 마이그레이션 도구입니다. Git이 코드 변경 이력을 관리하듯, DB 스키마의 변경 이력을 관리합니다.

### 4-3. 프로젝트의 Alembic 구성 요소

| 파일/폴더 | 역할 |
|---|---|
| `alembic.ini` | Alembic 전체 설정 파일 (마이그레이션 폴더 위치, 로그 설정 등) |
| `alembic/env.py` | 마이그레이션 실행 환경 설정. **DB 접속 URL과 모델 메타데이터를 연결**하는 핵심 파일 |
| `alembic/script.py.mako` | 마이그레이션 파일을 생성할 때 쓰는 **템플릿** |
| `alembic/versions/` | **생성된 마이그레이션 파일이 쌓이는 폴더** (아직 `revision`을 실행하지 않았다면 비어 있음) |
| `alembic/README.md` | `alembic init`이 만든 기본 안내 파일 |

> **`versions/`가 비어 보이는 이유:** 이 폴더는 `alembic revision` 명령을 실행해야 파일이 생깁니다. 모델을 작성하고 첫 revision을 만들면 `xxxx_메시지.py` 형태의 파일이 생성됩니다.

#### `alembic.ini` 주요 설정

| 설정 | 현재 값 | 설명 |
|---|---|---|
| `script_location` | `%(here)s/alembic` | 마이그레이션 스크립트 폴더 위치 (`ini` 파일이 있는 곳 기준 `alembic/`) |
| `prepend_sys_path` | `.` | 프로젝트 루트를 import 경로에 추가 (`app` 모듈을 찾기 위함) |
| `path_separator` | `os` | 경로 구분자 방식 (기본값) |
| `sqlalchemy.url` | 주석 처리됨 | DB URL을 여기 적지 않고 **`env.py`에서 앱 설정값으로 주입**하기 때문 |
| `[loggers]` 등 | 기본값 | 마이그레이션 실행 시 로그 출력 설정 |

#### `alembic/env.py` 코드별 설명

| 코드 | 설명 |
|---|---|
| `sys.path.append(os.getcwd())` | 현재 작업 폴더(프로젝트 루트)를 import 경로에 추가해 `app` 패키지를 불러올 수 있게 함 |
| `from app.core.db.databases import Base, DATABASE_URL` | 3번에서 만든 `Base`와 접속 URL을 그대로 재사용 |
| `from app import models` | 모델을 불러와 `Base.metadata`에 등록 (`app/models/__init__.py`가 import하는 모델들) |
| `config.set_main_option("sqlalchemy.url", DATABASE_URL)` | `alembic.ini`에는 없는 DB URL을 **앱의 `DATABASE_URL`로 주입** |
| `target_metadata = Base.metadata` | Alembic이 **현재 모델 구조와 실제 DB를 비교할 기준**. autogenerate의 핵심 |
| `run_migrations_offline()` | DB에 접속하지 않고 **SQL 문만 출력**하는 모드 (`--sql` 옵션) |
| `run_migrations_online()` | 실제 DB에 접속해 마이그레이션을 실행하는 모드. `async_engine_from_config`로 **비동기 엔진** 사용 |
| `poolclass=pool.NullPool` | 마이그레이션은 한 번 실행하고 끝나므로 커넥션 풀을 쓰지 않음 |
| `connection.run_sync(do_run_migrations)` | 비동기 연결 위에서 동기 방식의 Alembic 마이그레이션 함수를 실행 |
| `asyncio.run(run_migrations_online())` | 비동기 함수를 실행하는 진입점 |

> `env.py`가 비동기 방식(`async_engine_from_config`)으로 작성된 이유는 앱이 `asyncmy` 비동기 드라이버를 쓰기 때문입니다. (`alembic init -t async`로 만든 템플릿과 같은 구조)

### 4-4. 마이그레이션 수행 순서

> 모든 명령어는 **`alembic.ini`가 있는 프로젝트 루트**에서 실행합니다.

**① 의존성 설치** (`pyproject.toml`에 이미 `alembic`, `asyncmy`, `sqlalchemy[asyncio]`가 포함됨)

```bash
uv sync
```

**② `.env` 파일 생성** (3-1 참고)

**③ MySQL 컨테이너 실행**

```bash
docker compose up -d mysql
```

> Alembic은 **내 컴퓨터(호스트)에서** 실행하므로 `.env`의 `DB_HOST`는 `localhost`, `DB_PORT`는 docker-compose에서 열어둔 포트여야 합니다. (`docker-compose.yml`의 FastAPI 컨테이너는 `./app`만 마운트하므로 `alembic/`, `alembic.ini`가 컨테이너 안에 없습니다.)

**④ 모델 작성 & 등록** (4-1 참고)

**⑤ 마이그레이션 파일 생성 (모델 변경 감지)**

```bash
uv run alembic revision --autogenerate -m "create users table"
```

→ `alembic/versions/` 안에 마이그레이션 파일이 생성됩니다. **생성된 파일을 열어 내용을 반드시 확인**하세요.

**⑥ DB에 적용**

```bash
uv run alembic upgrade head
```

→ MySQL에 테이블이 생성되고, `alembic_version` 테이블에 현재 버전이 기록됩니다.

**⑦ 모델을 수정했다면 ⑤~⑥을 반복**

### 4-5. 생성되는 마이그레이션 파일 구조 (예시)

```python
"""create users table

Revision ID: a1b2c3d4e5f6
Revises:
Create Date: 2026-10-23 12:00:00
"""
from alembic import op
import sqlalchemy as sa

revision = "a1b2c3d4e5f6"
down_revision = None        # 이전 버전 (첫 마이그레이션이면 None)


def upgrade() -> None:
    # alembic upgrade 시 실행: 테이블/컬럼 생성
    op.create_table(
        "users",
        sa.Column("uuid", sa.CHAR(36), primary_key=True),
        sa.Column("email", sa.String(255), nullable=False),
        # ...
    )


def downgrade() -> None:
    # alembic downgrade 시 실행: upgrade의 반대 동작
    op.drop_table("users")
```

| 요소 | 설명 |
|---|---|
| `revision` | 이 마이그레이션의 고유 ID |
| `down_revision` | 바로 이전 마이그레이션 ID. 이 값들이 연결되어 **버전 체인**을 이룸 |
| `upgrade()` | 변경을 DB에 적용 |
| `downgrade()` | 변경을 되돌림 |

### 4-6. 자주 쓰는 명령어

| 명령어 | 설명 |
|---|---|
| `uv run alembic revision --autogenerate -m "메시지"` | 모델 변경을 감지해 마이그레이션 파일 생성 |
| `uv run alembic upgrade head` | 최신 버전까지 적용 |
| `uv run alembic downgrade -1` | 한 단계 되돌리기 |
| `uv run alembic current` | 현재 DB에 적용된 버전 확인 |
| `uv run alembic history` | 마이그레이션 이력 조회 |

### 4-7. 전체 흐름 요약

```
모델 작성/수정 (app/models) + __init__.py에 import
   → alembic revision --autogenerate  (변경 감지, alembic/versions/에 파일 생성)
   → 생성된 파일 검토
   → alembic upgrade head  (DB 반영)
```

### 4-8. 자주 만나는 오류

| 증상 | 원인과 해결 |
|---|---|
| `ModuleNotFoundError: No module named 'app'` | 프로젝트 루트가 아닌 곳에서 실행함. `alembic.ini`가 있는 폴더에서 실행 |
| 마이그레이션 파일이 비어 있음 (`pass`만 있음) | 모델이 `app/models/__init__.py`에서 import되지 않았거나 `Base`를 상속하지 않음 |
| `Can't connect to MySQL server` | MySQL 컨테이너가 꺼져 있거나 `.env`의 `DB_HOST`/`DB_PORT`가 다름 |
| `Target database is not up to date` | 적용 안 된 마이그레이션이 남아 있음. 먼저 `alembic upgrade head` 실행 |

> **주의:** autogenerate는 완벽하지 않으므로(컬럼 이름 변경 등은 삭제+추가로 인식), 생성된 마이그레이션 파일을 반드시 직접 확인한 뒤 적용하세요.

### [1] `app/core/config.py`

**의미:** `.env` 또는 환경변수에서 DB 접속 정보를 읽어 `settings` 객체로 제공합니다. 기본값은 값이 없을 때의 대체값입니다.

### [2] `app/core/db/databases.py`

**의미:** `settings`로 접속 URL을 만들고, 비동기 엔진·세션 팩토리·`Base`·`async_get_db()`를 정의합니다. 다른 모든 DB 관련 코드의 출발점입니다.

```python
from sqlalchemy.ext.declarative import declarative_base
from sqlalchemy.ext.asyncio import AsyncSession, create_async_engine, async_sessionmaker
from typing import AsyncGenerator
from app.core.config import settings


DATABASE_PREFIX = "mysql+asyncmy://"
DATABASE_URI = f"{settings.DB_USER}:{settings.DB_PASSWORD}@{settings.DB_HOST}:{settings.DB_PORT}/{settings.DB_NAME}"
DATABASE_URL = f"{DATABASE_PREFIX}{DATABASE_URI}"

# 비동기 엔진 생성
async_engine = create_async_engine(DATABASE_URL, echo=False, future=True)
# 비동기 세션 팩토리 생성
AsyncSessionLocal = async_sessionmaker(bind=async_engine, autoflush=False, expire_on_commit=False)
# 모델 베이스 생성
Base = declarative_base()

# 세션 생성 함수
async def async_get_db() -> AsyncGenerator[AsyncSession, None]:
    async with AsyncSessionLocal() as db:
        yield db
```

### [3] `app/core/db/models.py`

**의미:** 테이블 모델이 공통으로 사용하는 컬럼 묶음(Mixin)입니다. 실제 테이블은 `app/models/`에서 이 Mixin과 `Base`를 상속해 만듭니다.

```python
import uuid as uuid_pkg
from datetime import UTC, datetime

from sqlalchemy import Boolean, DateTime, text
from sqlalchemy.dialects.mysql import CHAR
from sqlalchemy.orm import Mapped, mapped_column
from uuid6 import uuid7


class UUIDMixin:
    uuid: Mapped[uuid_pkg.UUID] = mapped_column(
        CHAR(36), primary_key=True, default=uuid7
    )


class TimestampMixin:
    created_at: Mapped[datetime] = mapped_column(
        DateTime, default=datetime.now(UTC), server_default=text("current_timestamp(0)")
    )
    updated_at: Mapped[datetime | None] = mapped_column(
        DateTime, nullable=True, onupdate=datetime.now(UTC), server_default=text("current_timestamp(0)")
    )


class SoftDeleteMixin:
    deleted_at: Mapped[datetime | None] = mapped_column(DateTime, nullable=True)
    is_deleted: Mapped[bool] = mapped_column(Boolean, default=False)
```

### [5] `alembic/env.py`

**의미:** `alembic` 명령 실행 시 가장 먼저 읽히는 스크립트입니다. 앱의 `DATABASE_URL`과 `Base.metadata`를 연결하고, 비동기 엔진으로 마이그레이션을 실행합니다.

```python
import asyncio
from logging.config import fileConfig

from sqlalchemy import pool
from sqlalchemy.ext.asyncio import async_engine_from_config

from alembic import context

# 프로젝트 루트를 sys.path에 추가하여 app 모듈을 임포트할 수 있게 함
import os
import sys
sys.path.append(os.getcwd())

from app.core.db.databases import Base, DATABASE_URL

# Model Import
from app import models

# Alembic Config object
config = context.config

# Alembic 설정 파일의 sqlalchemy.url을 우리 앱의 DATABASE_URL로 덮어씁니다.
config.set_main_option("sqlalchemy.url", DATABASE_URL)

# Interpret the config file for Python logging.
# This line sets up loggers basically.
if config.config_file_name is not None:
    fileConfig(config.config_file_name)

# model's MetaData object
target_metadata = Base.metadata


def run_migrations_offline() -> None:
    """Run migrations in 'offline' mode.

    This configures the context with just a URL
    and not an Engine, though an Engine is acceptable
    here as well.  By skipping the Engine creation
    we don't even need a DBAPI to be available.

    Calls to context.execute() here emit the given string to the
    script output.
    """
    url = config.get_main_option("sqlalchemy.url")
    context.configure(
        url=url,
        target_metadata=target_metadata,
        literal_binds=True,
        dialect_opts={"paramstyle": "named"},
    )

    with context.begin_transaction():
        context.run_migrations()


def do_run_migrations(connection):
    context.configure(connection=connection, target_metadata=target_metadata)

    with context.begin_transaction():
        context.run_migrations()


async def run_migrations_online() -> None:
    """Run migrations in 'online' mode.

    In this scenario we need to create an Engine
    and associate a connection with the context.

    """
    connectable = async_engine_from_config(
        config.get_section(config.config_ini_section, {}),
        prefix="sqlalchemy.",
        poolclass=pool.NullPool,
    )

    async with connectable.connect() as connection:
        await connection.run_sync(do_run_migrations)

    await connectable.dispose()


if context.is_offline_mode():
    run_migrations_offline()
else:
    asyncio.run(run_migrations_online())
```

### [7] `docker-compose.yml`

**의미:** FastAPI 컨테이너와 MySQL 8.0 컨테이너를 정의합니다. MySQL의 DB 이름·사용자·비밀번호는 `.env`의 `${DB_NAME}`, `${DB_USER}` 등을 사용하고, 데이터는 `mysql_volume`에 저장되어 컨테이너를 지워도 유지됩니다.

```yaml
services:
  fastapi:
    build:
      context: .
      dockerfile: app/Dockerfile
    entrypoint: sh -c "uvicorn app.main:app --host 0.0.0.0 --port 8000 --reload"
    ports:
      - "8000:8000"
    volumes:
      - ./app:/app
      - static_volume:/app/static
      - media_volume:/app/media
    healthcheck:
      test: ["CMD-SHELL", "curl -f http://localhost:8000/healthcheck || exit 1"]
      interval: 5s
      timeout: 5s
      retries: 3
    depends_on:
      - mysql

  mysql:
    image: mysql:8.0
    volumes:
      - mysql_volume:/var/lib/mysql
    environment:
      MYSQL_ROOT_PASSWORD: ${DB_ROOT_PASSWORD}
      MYSQL_DATABASE: ${DB_NAME}
      MYSQL_USER: ${DB_USER}
      MYSQL_PASSWORD: ${DB_PASSWORD}
      MYSQL_PORT: ${DB_PORT}
    ports:
      - "${DB_PORT}:3306"
    healthcheck:
      test: ["CMD", "mysqladmin" ,"ping", "-h", "localhost"]
      timeout: 20s
      retries: 5


volumes:
  mysql_volume:
  static_volume:
  media_volume:
```
